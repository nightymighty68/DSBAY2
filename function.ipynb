```python
A =[56,25,36,89,95,23,67,45,34,18]
```

```python
for i in range(len(A)):
    for j in range(i+1,len(A)):
        if(A[i]>A[j]):
            temp=A[i]
            A[i]=A[j]
            A[j]=temp
```
